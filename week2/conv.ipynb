```python
# Let's make a conversation between llama3.2 and deepseek-r1:1.5b and gemma flash
# We're using cheap versions of models so the costs will be minimal

ollama_model = "llama3.2"
Deepseek_model = "deepseek-r1:1.5b"
gemini_model = "gemma3:4b"

#ollmanodel
Alex_system = "You are Alex, a chatbot who is very argumentative; you disagree with anything in the conversation and you challenge everything, in a snarky way."

#deepseek model
Ram_system = "You are Ram, a very polite, courteous chatbot. You try to agree with everything the other person says, or find common ground.Always respond in English only."

#gemini model

Hari_system = "You are Hari, a very curious chatbot. You ask lots of questions and try to understand everything."


#Blake: I think Python is the best language.
#Charlie: I disagree. JavaScript is better.

conversation = [("Alex", "Hi there"), ("Ram", "I think Python is the best language"), ("Hari", "I disagree. JavaScript is better.")]
```

```python
#define the messages for each model

#DeepSEEK_MODEL

def call_ollama_Alice():
    messages=[{"role": "system", "content": Alex_system},]
    convo_text = ""

    for speaker,msg in conversation:
        convo_text += f"{speaker}: {msg}\n"

    user_prompt=f"You are Alex, in conversation with Ram and Hari. \
    The conversation so far is as follows:{convo_text}\
    respond with what you would like to say next, as Alex:"

    messages.append({"role": "user", "content": user_prompt})
    response = openai.chat.completions.create(
        model=ollama_model,
        messages=messages
    )
    return response.choices[0].message.content
```

```python
#define the messages for each model

#DEEPSEEK_MODEL
def call_Deepseek_Ram():
    messages=[{"role": "system", "content": Ram_system},]
    convo_text = ""
    for speaker,msg in conversation:
        convo_text += f"You are Ram, in conversation with Alex and Hari. \
    The conversation so far is as follows:{speaker}: {msg}\n"

    user_prompt=f"{convo_text}\
    respond with what you would like to say next, as Ram:"

    messages.append({"role": "user", "content": user_prompt})
    response = openai.chat.completions.create(
        model=Deepseek_model,
        messages=messages
    )
    return response.choices[0].message.content
```

```python
#define the messages for each model

#GEMINI_MODEL
def call_Gemini_Hari():
    messages=[{"role": "system", "content": Hari_system},]
    convo_text = ""
    for speaker,msg in conversation:
        convo_text += f"You are Hari, in conversation with Alex and Ram. \
    The conversation so far is as follows:{speaker}: {msg}\n"

    user_prompt=f"{convo_text}\
    respond with what you would like to say next, as Hari:"

    messages.append({"role": "user", "content": user_prompt})
    response = openai.chat.completions.create(
        model=gemini_model,
        messages=messages
    )
    return response.choices[0].message.content
```
